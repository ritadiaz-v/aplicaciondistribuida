```python
import socket
import threading
import time

# Función del servidor
def servidor():
    servidor_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    servidor_socket.bind(('localhost', 12345))
    servidor_socket.listen(1)
    print("🟢 Servidor: esperando conexión...")

    conexion, direccion = servidor_socket.accept()
    print(f"🟢 Servidor: conectado con {direccion}")

    mensaje = conexion.recv(1024).decode()
    print("🟢 Servidor recibió:", mensaje)

    respuesta = "Hola cliente, soy el servidor."
    conexion.send(respuesta.encode())

    conexion.close()
    print("🟢 Servidor: conexión cerrada")

# Función del cliente
def cliente():
    time.sleep(1)  # Espera para que el servidor esté listo

    cliente_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    cliente_socket.connect(('localhost', 12345))

    mensaje = "Hola servidor, soy el cliente."
    cliente_socket.send(mensaje.encode())
    print("🔵 Cliente envió:", mensaje)

    respuesta = cliente_socket.recv(1024).decode()
    print("🔵 Cliente recibió:", respuesta)

    cliente_socket.close()
    print("🔵 Cliente: conexión cerrada")

# Crear hilos
hilo_servidor = threading.Thread(target=servidor)
hilo_cliente = threading.Thread(target=cliente)

# Iniciar hilos
hilo_servidor.start()
hilo_cliente.start()

# Esperar que terminen
hilo_servidor.join()
hilo_cliente.join()
```
